# Quantum Entanglement: A Joint State That Cannot Be Separated

A joint pure state is entangled when it cannot be written as a product of subsystem states. The Bell state |Φ+⟩=(|00⟩+|11⟩)/√2 is entangled: its two amplitudes describe a coherent joint state. Measuring both qubits in Z gives 00 or 11, each with probability 1/2.

The classical mixture ρ_mix=(|00⟩⟨00|+|11⟩⟨11|)/2 has the same Z statistics, but lacks the off-diagonal coherence of |Φ+⟩⟨Φ+|. Z correlations alone therefore do not establish entanglement. For the Bell state, XX and ZZ correlations are +1, while YY is −1. For this mixture, XX and YY are zero and ZZ remains +1.

The companion uses ρ(v)=v|Φ+⟩⟨Φ+|+(1−v)ρ_mix for 0≤v≤1. Its same-basis correlation is C_Z=1, C_X=v, or C_Y=−v. Each local outcome is unbiased. The four joint probabilities are P(++ )=P(−−)=(1+C)/4 and P(+−)=P(−+)=(1−C)/4, with 0/1 replacing +/− for Z.

Tracing out either qubit gives I/2 for every v in this family. An observer cannot use local outcomes to learn the distant choice or send a message. A Bell-test claim additionally requires suitable measurement settings, assumptions, statistics, and experimental controls. This limited model teaches coherence and correlations; it does not by itself implement a loophole-free Bell test.

## Predict before running

At v=1 in X, ++ and −− occur with probability 1/2 each. At v=0 in X, all four outcomes occur with probability 1/4. Both cases produce 00 and 11 equally in Z. In Y, Φ+ instead gives opposite outcomes with certainty.

```python
from pathlib import Path
import sys, math
base=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(base))
import foundations as f
```

```python
for v in (0,.5,1):
    for basis in ('X','Y','Z'):
        result=f.bell(v,basis)
        assert abs(sum(result['probabilities'])-1)<1e-12
        assert result['local_probability']==.5
        print(v,basis,result)
assert f.bell(1,'Y')['correlation']==-1
assert f.bell(0,'X')['correlation']==0
```
