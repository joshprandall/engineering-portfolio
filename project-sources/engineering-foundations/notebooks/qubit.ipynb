# Qubits and Superposition Without the Mysticism

A pure qubit is a normalized vector α|0⟩+β|1⟩ with complex amplitudes and |α|²+|β|²=1. Measurement in the computational Z basis produces 0 with probability |α|² and 1 with probability |β|². These probabilities do not reveal every feature of the state.

Ignoring an unobservable global phase, write |ψ⟩=cos(θ/2)|0⟩+exp(iφ)sin(θ/2)|1⟩. The angle θ ranges from 0 to π; φ describes relative phase. The corresponding Bloch vector is (sinθ cosφ, sinθ sinφ, cosθ). An ideal pure state has Bloch-vector norm one.

The probability of the positive outcome along a Pauli axis is (1+r_axis)/2. Thus P(X=+)=(1+sinθ cosφ)/2, P(Y=+)=(1+sinθ sinφ)/2, and P(Z=0)=(1+cosθ)/2. Rotating relative phase changes interference in X and Y even when Z probabilities stay fixed.

At θ=π/2, φ=0 the state is |+⟩: X gives + with certainty, while Z is balanced. At φ=π/2 the state is (|0⟩+i|1⟩)/√2 and Y gives + with certainty. Multiplying both amplitudes by the same complex phase changes none of these predictions. These calculations concern a pure, ideal qubit; mixed states and hardware errors require additional descriptions.
