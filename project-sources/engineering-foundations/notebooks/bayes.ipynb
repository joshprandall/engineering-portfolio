# Bayes’ Theorem

Conditional probability changes the population over which a fraction is calculated. P(A|B) asks for the fraction of B cases that also satisfy A; P(B|A) asks the reverse question. These are generally different. From P(A∩B)=P(A|B)P(B)=P(B|A)P(A), Bayes’ rule follows: P(A|B)=P(B|A)P(A)/P(B), provided P(B)>0.

Let A mean a machine truly has a fault and B mean a detector raises an alarm. In a fictional population of 10,000 machines with a 1% fault rate, 100 have the fault. With 90% sensitivity, 90 of those machines alarm. Of the 9,900 healthy machines, a 5% false-positive rate produces 495 false alarms. Therefore only 90/(90+495)≈15.38% of alarms identify true faults.

The denominator is the complete positive population. It combines true positives and false positives: P(B)=P(B|A)P(A)+P(B|not A)P(not A). Ignoring the healthy machines replaces the requested conditional probability with sensitivity. The answer changes when prevalence changes even if detector behavior stays fixed.

These are expected counts using assumed rates, not observations from a real deployment. Rates estimated from finite data have uncertainty and may drift across environments. Repeated tests cannot be multiplied as independent evidence unless their conditional dependence is justified. A useful investigation records assumptions before interpreting a posterior.

## Predict before running
