# Subnetting Without Memorization

An IPv4 address has 32 bits. A /26 prefix fixes the first 26 and leaves six that can vary. Six bits encode 2⁶ = 64 combinations. The network address is found by setting those six host bits to zero; setting them to one gives the upper address. Prefix notation identifies an aligned set of addresses, not a number of computers.

For 192.168.10.70/26, the last-octet blocks start at 0, 64, 128, and 192. Since 64 ≤ 70 < 128, the block is .64–.127. Equivalently, apply the mask 255.255.255.192. A host value cannot be used as an arbitrary block start.

For a conventional broadcast subnet with prefix at most /30, reserving the all-zero and all-one host patterns leaves 2^(32−prefix)−2 conventional host addresses. This usage rule is not the definition of CIDR. RFC 3021 permits both addresses on a /31 point-to-point link. A /32 denotes one address. Cloud platforms can reserve additional addresses, so their allocatable count may differ.

A routing prefix and a broadcast LAN need not describe the same operational object. Longest-prefix routing, address assignment, and platform reservations are related decisions with different rules. First calculate the block; then state the intended use before reporting a host count.
