# Eigenvalues and Eigenvectors

A matrix represents a linear transformation after choosing input and output bases. Multiplying A by a vector combines the columns of A using the vector components as weights. An eigenvector is a nonzero vector whose image lies on the same line: Av = λv. A negative eigenvalue reverses direction, and zero sends the vector to zero.

For A=[[2,1],[1,2]], multiplying (1,1) gives (3,3), so λ=3 is an eigenvalue with eigenvector (1,1). Multiplying (1,−1) gives (1,−1), so λ=1 is another eigenvalue. These are statements about directions, not unique vector lengths: any nonzero scalar multiple is also an eigenvector.

To find candidates algebraically, solve det(A−λI)=0. Here (2−λ)²−1 = λ²−4λ+3 = (λ−1)(λ−3), giving 1 and 3. Substitute into (A−λI)v=0 to obtain the corresponding directions. For distinct eigenvalues these two directions are independent; they form a basis in which this transformation is diagonal. Not every matrix has a complete eigenbasis.

A computation should report the residual r=Av−λv. An absolute residual depends on the size of v, so compare ||r||/||v|| and, for broader numerical work, scale by the matrix norm as well. A tiny residual checks the submitted pair; it does not prove that all eigenpairs were found or that an ill-conditioned problem is harmless.

## Predict before running

A(1,1)=(3,3)=3(1,1), giving residual zero. If the same vector is paired with λ=2, r=(1,1), ||r||=√2 and ||r||/||v||=1. Scaling the vector by 100 scales the absolute residual by 100 but leaves the latter ratio unchanged.

```python
result = f.eigen()
assert result['residual'] == 0
assert f.eigen(vector=(1,-1),eigenvalue=1)['residual'] == 0
wrong = f.eigen(vector=(1,-1),eigenvalue=3)
assert abs(wrong['residual'] - math.sqrt(8)) < 1e-12
print('Correct:', result, '\nIncorrect candidate:', wrong)
```
