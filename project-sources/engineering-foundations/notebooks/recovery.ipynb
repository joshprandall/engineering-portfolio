# RTO and RPO: Two Different Recovery Targets

Recovery Time Objective (RTO) and Recovery Point Objective (RPO) constrain different consequences of an outage. RTO is a target for how quickly the service must recover under a defined incident clock. RPO specifies the recoverable data point needed after an outage, often expressed as the maximum tolerable age or loss interval. The business impact analysis must define scope and clock boundaries.

In a fictional incident starting at 14:00, an RTO of 120 minutes requires the defined service to recover by 16:00. An RPO of 30 minutes requires a recoverable data point at least as recent as 13:30. Restoring at 15:35 from a verified 13:15 point meets the time objective but misses the data objective by 15 minutes. One result cannot substitute for the other.

Backup-job success is an observation about a process. Demonstrated recovery requires evidence that the selected data is recoverable, the application starts, dependencies are available, identities and permissions work, and the required business checks pass. A recovered file count alone cannot establish transactional or application consistency.

Record the incident start, chosen restore point, restoration completion, validation result, dependency assumptions, and who approved the exercise outcome. A successful exercise supports that tested scenario; it does not guarantee the same result at another scale or during a different failure. Improve the runbook by investigating the limiting objective, not by changing the target after seeing the result.
