# SLI, SLO, SLA

A Service Level Indicator (SLI) is a defined measurement of service behavior. A Service Level Objective (SLO) is a target for that indicator over a specified window. A Service Level Agreement (SLA) is an agreement with consequences and terms; it is not automatically created by publishing an internal target.

For a request-based success SLI, define eligible requests and bad outcomes before counting. Observed success is 1−bad/eligible. A 99.9% SLO permits a bad fraction of 0.001. Over 1,000,000 eligible requests that is 1,000 bad requests. If 1,500 are bad, observed success is 99.85%, the remaining budget is −500, and consumption is 150%.

Request-based and time-based indicators answer different questions. A failed request is not necessarily one minute of downtime. Traffic variation, exclusions, retries, partial failures, missing telemetry, and the measurement window all affect interpretation. A service that fails during peak demand can consume a request budget very differently from one that fails during quiet hours.

An error-budget policy connects measurements to pre-agreed actions: investigate, pause selected risky changes, improve resilience, or explicitly accept a tradeoff. The arithmetic does not choose the business policy. An incident review should distinguish the observed impact, contributing technical factors, decisions under uncertainty, and follow-up work with owners and verification criteria. Avoid changing eligibility rules after an incident merely to make the metric pass.
