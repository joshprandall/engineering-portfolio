# Measure phase error on a circle

Prerequisites: Phase kickback, the inverse QFT and probability-weighted error.

After ideal kickback, control amplitude k is exp(2πiφk)/√N. Applying the inverse DFT gives the QPE output amplitudes. The notebook calculates that matrix-vector product independently of the sine-quotient formula in the existing browser lab. Exact grid phases yield one ideal output bin; off-grid phases distribute probability across several bins.

Phase is periodic. Near φ=0.99, a reported phase of 0 is only 0.01 away around the circle, although ordinary subtraction gives −0.99. Circular error wraps the difference into [−1/2,1/2). The notebook reports both linear and circular RMSE, each weighted by the exact probability distribution. The browser retains its existing linear sample statistic and adds the circular sample statistic.

Exact probability-weighted RMSE and finite-shot sample RMSE are different quantities. Seed, shot count, bit order and any mixture parameter belong in the exported result. A uniform probability mixture is a teaching approximation, not calibrated device noise. The simulator does not implement controlled powers of a supplied Hamiltonian or demonstrate hardware advantage.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Resolution | 1/2ⁿ | Grid spacing, not guaranteed error |
| Linear error | y/N−φ | Treats endpoints as far apart |
| Circular error | wrap(y/N−φ) | Respects periodic phase |

## Predict

For φ=1/8 and n=3, bin 001 has probability one. For φ=0.99, compare linear and circular RMSE; the circular result should be smaller.

```python
from pathlib import Path
import sys, math
base=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(base))
import quantum_models as m
```

```python
result=m.qpe(.125,3)
assert result['dominant']==1
assert abs(result['probabilities'][1]-1)<1e-12
wrapped=m.qpe(.99,3)
assert wrapped['circular_rmse']<wrapped['linear_rmse']
print(result)
print('Near wrap:',wrapped)
```
