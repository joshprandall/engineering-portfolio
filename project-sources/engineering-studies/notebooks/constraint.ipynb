# Keep learned scores inside explicit constraints

Prerequisites: Binary choices, objective functions and feasibility.

A predicted score and a feasible decision are different objects. Three candidate choices have scores 0.8,0.6,0.7. At most two may be selected, and choices 0 and 2 cannot coexist. Thresholding all scores at 0.5 selects all three and violates both constraints.

The companion enumerates all eight assignments, removes infeasible ones and maximizes the score sum over the remainder. It chooses (1,1,0) with objective 1.4. This small exact solver demonstrates the interface between uncertain learned preferences and explicit symbolic constraints. The scores are fixtures, not a trained or calibrated model.

Feasibility is only as meaningful as the encoded constraints. A perfectly solved incomplete model can still produce an unacceptable real-world decision. For a larger problem, choose a suitable SAT/SMT, integer-programming or constraint solver and validate its translation. Record unsatisfiable inputs distinctly from an implementation failure or a low-quality feasible optimum.

## Model structure

| Element | Model | Interpretation |

| Constraints | Capacity ≤2; not both 0 and 2 | Feasible set |

## Predict

At capacity two, the best feasible selection is choices 0 and 1. At capacity zero the all-zero assignment remains feasible, with objective zero.
