# Measure product-formula error

Prerequisites: Matrix exponentials, noncommuting operators and operator norms.

For H=X+Z, the exact propagator is exp(−itH). Because X and Z do not commute, exp(−itX)exp(−itZ) is generally different. The companion splits time into r steps and compares [exp(−itX/r)exp(−itZ/r)]ʳ with an exact eigen-decomposition reference.

Both propagators are unitary to rounding error, so normalization alone cannot show that the approximate dynamics are accurate. Measure the operator-norm difference and inspect refinement at fixed total time. This first-order product formula is a small reference calculation, not a gate-count estimate for arbitrary many-body simulation.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Target | exp(−it(X+Z)) | Exact dynamics |
| Approximation | Repeated separated X and Z steps | Noncommutation creates error |
| Checks | Accuracy and unitarity | Different properties |
