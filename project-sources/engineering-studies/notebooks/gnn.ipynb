# Test message passing under node relabeling

Prerequisites: Adjacency matrices, vector features, permutations and pooling.

On a three-node path, use h=(1,2,4) and update h′=ReLU(0.5h+Ah). The shared message rule sums neighboring features and combines them with a scaled self feature. This gives h′=(2.5,6,4). Summation does not depend on the order in which neighbors are listed.

Relabeling nodes by a permutation P changes features to Ph and adjacency to PAPᵀ. A permutation-equivariant layer must then return Ph′. A pooled sum is invariant because relabeling merely reorders its terms. These are separate properties: node outputs move with nodes, while the graph-level summary stays the same. Changing features without also permuting adjacency is a different graph-input pair, not a valid symmetry test.

The companion uses fixed weights and a tiny graph; it is an executable layer, not a trained GNN benchmark. Deeper message passing can mix information until nodes become hard to distinguish. Testing one permutation establishes one numerical check, while the algebraic form explains why the property holds for every consistent relabeling.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Node update | h′=ReLU(0.5h+Ah) | Shared local rule |
| Relabeling | A→PAPᵀ, h→Ph | Transform both inputs |
| Graph output | Σᵢh′ᵢ=12.5 | Invariant pooled scalar |

## Predict
