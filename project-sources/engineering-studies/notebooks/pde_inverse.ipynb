# Separate an exact diffusion mode from an inverse problem

Prerequisites: Differential equations, eigenfunctions, likelihoods and normalization.

On a sphere of radius r, y/r is a degree-one spherical harmonic with Laplace–Beltrami eigenvalue −2/r². Therefore u(y,t)=0.5+0.5(y/r)exp(−2Dt/r²) solves the surface heat equation for this particular initial field. The surface mean stays 0.5 and the north–south contrast decays exponentially. This analytic example is not a numerical solver for arbitrary geometry or arbitrary initial conditions.

The inverse companion uses fictional curvature observations 0.98,1.04,1.01 with independent Gaussian noise σ=0.05 and a uniform radius prior on [0.4,2.5]. A discrete radius grid receives weights proportional to exp[−Σ(Kᵢ−1/r²)²/(2σ²)]. Subtracting the largest log weight before exponentiating avoids unnecessary underflow without changing normalized probabilities.

A narrow posterior can still be wrong if observations are correlated, σ is misspecified, or the prior truncates the plausible radius. The equal-tail interval is a posterior statement under this model, not a universal guarantee about repeated measurements. Compare grids before interpreting digits beyond their spacing.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Forward model | One spherical harmonic | Exact exponential decay |
| Likelihood | Independent Gaussian curvature noise | Explicit uncertainty assumption |
| Posterior | 501 normalized grid weights | Resolution and prior limits |
