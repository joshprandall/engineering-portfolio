# Bound memory by consent, purpose and expiry

Prerequisites: Data minimization, explicit state and expiration semantics.

A personal assistant needs a policy for what it retains, why it retains it and when the data stops being eligible for use. The fixture stores a fictional display preference only when consent is true and the current time precedes the expiry time. At expiry it returns no active memory.

This is an ephemeral policy calculation, not a user account or a persistent memory service. It does not infer consent from the text of a request or quietly keep a second copy. A real design must propagate deletion and expiry through caches, indexes, logs and backups according to its declared policy. Test denied consent and exact expiry boundaries as carefully as the successful path.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Consent | Explicit boolean | Gate retention |
| Purpose | Display preference | Bound use |
| Expiry | now < expiry | Reject at the boundary |

## Predict

With consent=false, memory is empty. With consent=true, now=10, expiry=20 it is available; at now=20 it is no longer active.
