# Decode bit-flip syndromes and expose the limit

Prerequisites: Pauli operators, parity and independent error probabilities.

The three-qubit repetition code encodes |0⟩ as |000⟩ and |1⟩ as |111⟩. Stabilizers Z₀Z₁ and Z₁Z₂ reveal parity differences without directly measuring the encoded logical bit. A single X error on qubit 0,1,2 produces syndrome 10,11,01 respectively, which identifies a corrective X in the single-error model.

Under independent bit-flip probability p, a majority decoder fails when two or three qubits flip: 3p²(1−p)+p³=3p²−2p³. The notebook enumerates all eight error patterns, weights their probabilities and verifies that polynomial. Double errors can share a syndrome with a single error and be miscorrected, so a syndrome is not a uniquely identified arbitrary error.

The code does not protect an arbitrary logical state against phase errors. A phase flip can act logically without changing these Z-type check outcomes. General quantum error correction needs complementary information and a noise-aware fault-tolerant measurement procedure; this perfect-syndrome toy does not establish a surface-code threshold.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Checks | Z₀Z₁, Z₁Z₂ | Two parity bits |
| Correctable model | At most one X | Distinct nonzero syndromes |
| Failure | Two or three X errors | 3p²−2p³ |
