# Check synchronization against numerical resolution

Prerequisites: Ordinary differential equations, graphs, phases and numerical integration.

The model evolves phase θᵢ with intrinsic frequency ωᵢ and degree-normalized coupling KΣⱼAᵢⱼsin(θⱼ−θᵢ)/max(1,dᵢ). The order parameter R=|N⁻¹Σᵢexp(iθᵢ)| lies between zero and one and measures instantaneous phase concentration. It is not a measure of intelligence, a physical universe, or proof that every oscillator is frequency-locked.

The companion creates a fixed seeded random graph, frequencies and initial phases, then runs explicit Euler integration to the same final time with steps Δt and Δt/2. It compares phases using wrapped differences and reports both R values. Holding the graph and initial condition fixed isolates discretization effects; changing the random seed at the same time would confound that comparison.

One refinement comparison is a diagnostic rather than a convergence proof. Repeat another halving and then an ensemble of seeds before claiming a stable regime.
