# Fit a denoiser at one noise level

Prerequisites: Gaussian variables, regression and train/test separation.

The forward-noising equation is xₜ=√ᾱx₀+√(1−ᾱ)ε. Here x₀ and ε are independent standard Gaussians. The optimal conditional-mean predictor is √ᾱxₜ, with expected squared error 1−ᾱ. The notebook fits a scalar coefficient on 2,000 training pairs and evaluates it on 2,000 different pairs.

This demonstrates one learnable denoising problem and an analytic baseline. It is not a multi-step reverse diffusion sampler or an image generator. Knowing the injected noise would make perfect algebraic reconstruction possible, but a real denoiser must estimate information that is not supplied. Compare the fitted coefficient and held-out error with the analytic values before extending to a schedule of noise levels.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Forward | Signal + independent noise | Known sampling distribution |
| Fit | One coefficient from training pairs | Learn a conditional predictor |
| Evaluate | Held-out MSE against 0.6 | Sampling variability remains |

## Predict
