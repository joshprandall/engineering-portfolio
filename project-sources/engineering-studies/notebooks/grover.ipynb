# Observe amplification and overshooting

Prerequisites: Normalized state vectors, reflections and amplitude versus probability.

With one marked state among N possibilities, initialize every amplitude to 1/√N. The oracle changes the marked amplitude's sign, and diffusion reflects all amplitudes about their mean. Repeating this pair rotates the state in the plane spanned by marked and unmarked components.

Let θ=arcsin(1/√N). After k iterations the marked probability is sin²((2k+1)θ). More iterations do not always improve success: the state can rotate past the marked direction. The companion applies the sign flip and mean reflection directly, checks normalization, and compares the result with the analytic formula.

The oracle here is an explicitly selected array index. Constructing an oracle for a real problem has a cost that this fixture does not model. The dense classical simulation is an educational reference, not a search speedup. Multiple marked states, noisy gates and unknown solution count require a changed analysis.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Oracle | Flip one amplitude sign | Encodes the marked set |
| Diffusion | aᵢ→2mean(a)−aᵢ | Reflection |
| Iteration count | sin²((2k+1)θ) | Oscillatory success |
