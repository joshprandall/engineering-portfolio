# Solve a small game exactly before learning it

Prerequisites: Turn-based games, terminal states and dynamic programming.

In this normal-play subtraction game, a player removes one or two stones and the player taking the last stone wins. Position zero is losing for the player to move. A positive position is winning if any legal move leads to a losing position for the opponent. The companion builds this table from small positions upward.

For five stones, taking two leaves three, a losing position. This is exact dynamic programming for a fully known game, not an RL agent and not a change to either protected portfolio game. It provides a clean baseline for a later learned policy: compare decisions against the solved state table, including losing states where no winning move exists.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Terminal | Zero stones | Losing for the next player |
| Recurrence | Any move to a losing state? | Winning-state test |
| Policy | Select such a move | Exact model-based baseline |
