# Use T1 and T2 with an explicit noise convention

Prerequisites: Exponential decay, populations, coherence and time units.

In a simple Markovian relaxation-plus-dephasing model, excited-state population decays as exp(−t/T1), while off-diagonal coherence decays as exp(−t/T2). The rates obey 1/T2=1/(2T1)+1/Tφ, so nonnegative pure dephasing requires T2≤2T1. The notebook rejects parameter combinations outside that model.

These quantities are not interchangeable with gate fidelity, readout error, connectivity or logical error rate. Device modalities also have different control and engineering constraints. This companion uses arbitrary consistent time units and fictional parameters; it does not rank current hardware vendors or substitute historical roadmap claims for measured performance.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| T1 | Population relaxation | Excited-state survival |
| T2 | Coherence decay | Includes relaxation contribution |
| Tφ | Additional dephasing | Nonnegative rate |
