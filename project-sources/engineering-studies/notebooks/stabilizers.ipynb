# Check commuting Pauli parity measurements

Prerequisites: Tensor products, Pauli anticommutation and stabilizer syndromes.

The four-data-qubit toy uses checks XXXX, ZZII and IZZI. Each pair commutes: overlapping X and Z factors anticommute at two positions, yielding a positive net sign. An X error on qubit 1 anticommutes with both Z checks but commutes with the X check, producing syndrome (0,1,1).

This calculation teaches check compatibility and syndrome construction. It is not a full surface-code patch, repeated syndrome circuit or decoder. Surface-code analysis also needs boundaries, logical operators, data and measurement noise, code distance, decoding and resource accounting. A commuting set alone does not establish a threshold or fault-tolerant logical gate.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Checks | XXXX, ZZII, IZZI | Compatible Pauli observables |
| Error | IXII | One explicit data error |
| Syndrome | Commutation signs | Locates a check violation |
