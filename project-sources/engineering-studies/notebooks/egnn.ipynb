# Check an equivariant coordinate update

Prerequisites: Distances, rotations, translations and graph message passing.

A distance-based message can remain unchanged when all coordinates rotate or translate together. The companion uses weights exp(−||xᵢ−xⱼ||²) and an update x′ᵢ=xᵢ+0.1Σⱼ(xᵢ−xⱼ)wᵢⱼ. Distances are invariant; displacement vectors rotate; summing weighted displacements therefore produces a coordinate update that rotates and translates with the input.

This is a deliberately small EGNN-style coordinate rule inspired by the paper's architecture, with fixed functions rather than learned networks. Three two-dimensional points are transformed by a rotation and translation, processed, and compared with transforming the original outputs. The residual should be near machine precision. Translation cancels in pairwise differences and reappears through the original coordinate xᵢ.

Equivariance alone does not make a model accurate, stable across many layers, or appropriate for every physical quantity. A task may require conservation laws, reflection-sensitive features or boundary conditions that this update does not encode. A symmetry test should specify the transformation group and the type of each output.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Scalar message | Distance squared | Invariant to rigid transformations |
| Coordinate difference | xᵢ−xⱼ | Rotates as a vector |
| Output test | f(Qx+t)−Qf(x)−t | Near-zero residual |
