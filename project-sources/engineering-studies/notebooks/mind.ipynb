# Coordinate roles with testable artifacts

Prerequisites: Typed tools, dependency graphs, hashing and explicit acceptance criteria.

A coordinator should know which artifact satisfies each task dependency. The local MIND companion uses three deterministic roles. The planner records inputs (1,2,3) and expected sum 6. The executor computes the sum. The verifier compares the computed artifact with the explicit expected value. Each role's serialized artifact receives a SHA-256 digest in the event ledger.

This differs from counting evidence-looking words: a verifier must inspect the actual artifact and a discriminating criterion. A digest identifies bytes; it does not certify their truth or the competence of the author. Dependencies are checked before a role runs, and a JSON round trip demonstrates a serializable checkpoint. A failed validation produces blocked, not a decorated completion status.

The fixture represents coordination protocol, not autonomous LLM collaboration or distributed consensus. All roles run sequentially in one process. To extend it, introduce an artifact schema, injected failures, persisted attempts, cancellation and an idempotent executor. Separate a role's claim that it succeeded from the verifier's observable result.
