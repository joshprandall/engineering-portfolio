# Compare GHZ and product probes under dephasing

Prerequisites: Parameter estimation, Fisher information and noise assumptions.

For frequency estimation at fixed interrogation time t, this convention gives product-probe QFI Nt²exp(−2γt) and GHZ-probe QFI N²t²exp(−2Nγt). The GHZ phase accumulates N times faster, but independent dephasing also damps its coherence N times faster. In the noiseless limit the expressions recover N and N² scaling at fixed t.

The comparison fixes interrogation time; optimizing time with a fixed total experimental budget is a different resource question. QFI is an information bound, not a measured estimator variance. Preparation, readout, correlated noise and accessible measurements can further constrain a real sensor. A larger ideal expression alone does not establish a usable sensing advantage.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Signal | Frequency-dependent phase | Longer time increases sensitivity |
| Product probes | Nt²e^(−2γt) | Independent information |
| GHZ probes | N²t²e^(−2Nγt) | Enhanced signal and fragility |
