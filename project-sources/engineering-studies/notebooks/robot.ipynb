# Check a reachable target with forward kinematics

Prerequisites: Trigonometry, coordinate frames and workspace constraints.

For two planar unit links, the endpoint is (cos q₁+cos(q₁+q₂), sin q₁+sin(q₁+q₂)). The law of cosines gives cos q₂=(x²+y²−2)/2. A value outside [−1,1] means the requested point is unreachable under this geometry and must be rejected rather than clamped into a fabricated solution.

The companion chooses one elbow branch, solves q₁ and q₂, then independently recomputes the endpoint with forward kinematics. This is position geometry only: it includes no collision map, joint limits, torque, dynamics, feedback control or real robot connection. A reachable point is not automatically a safe motion. Plan a trajectory and validate the omitted constraints before considering physical actuation.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Inverse problem | Target → candidate angles | Choose an elbow branch |
| Forward check | Angles → endpoint | Measure residual |
| Missing constraints | Collision, limits, dynamics | No actuation claim |
