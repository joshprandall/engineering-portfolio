# Move an eigenphase onto an ancilla

Prerequisites: Tensor products, controlled gates, eigenvectors and the Hadamard gate.

If U|u⟩=e^(2πiφ)|u⟩, applying controlled-U to |+⟩|u⟩ gives (|0⟩+e^(2πiφ)|1⟩)|u⟩/√2. The target remains in its eigenstate while the control acquires a relative phase. That phase is global only within the target branch; it is relative between the two control branches and can therefore affect interference.

The companion explicitly constructs the two-qubit controlled matrix for U=diag(1,e^(2πiφ)), with the ancilla listed first and target |1⟩. A final Hadamard on the ancilla gives P(0)=cos²(πφ). Measuring the control directly in Z before that Hadamard would always give one half and miss the relative phase.

The eigenstate assumption matters. A superposed target can correlate the control with different eigenphases. QPE uses a larger control register and powers of U to resolve that information rather than inferring a unique phase from a single probability.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Prepare | |+⟩⊗|1⟩ | Control branches are coherent |
| Controlled U | Phase on the control-1 branch | Relative phase encodes φ |
| Interfere | Hadamard then measure | P(0)=cos²(πφ) |
