# Track a noisy state with a density operator

Prerequisites: Complex vectors, outer products, trace and eigenvalues.

A density operator represents both pure states and probabilistic mixtures. It must be Hermitian, positive semidefinite and trace one. For |+⟩ the matrix has four entries equal to 1/2; its purity Tr(ρ²) is one. A classical mixture of |0⟩ and |1⟩ instead has zero off-diagonal coherence and purity 1/2.

Amplitude damping uses K₀=diag(1,√(1−γ)) and K₁=[[0,√γ],[0,0]]. The channel sums KᵢρKᵢ†. For the initial |+⟩, the resulting matrix is [[(1+γ)/2,√(1−γ)/2],[√(1−γ)/2,(1−γ)/2]]. This is a specific relaxation channel, not a universal hardware noise model. At γ=1 it resets the state to pure |0⟩, illustrating that noise does not always monotonically reduce purity.

The notebook checks trace, eigenvalues and purity. A negative eigenvalue beyond numerical tolerance is a physical-validity error, not merely a strange probability plot. Compare channel composition with replacing amplitudes by classical probabilities; discarding off-diagonal terms removes measurable phase information.
