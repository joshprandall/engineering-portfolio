# Compare a variational energy with exact diagonalization

Prerequisites: Hermitian matrices, expectation values and constrained optimization.

Use H=Z⊗I+I⊗Z+0.5X⊗X and normalized ansatz |ψ(θ)⟩=cosθ|00⟩+sinθ|11⟩. Its energy is E(θ)=2cos(2θ)+0.5sin(2θ). The variational principle says a normalized state's energy cannot fall below the exact ground energy; a lower result indicates a convention, normalization or calculation problem.

The even-parity block is [[2,0.5],[0.5,−2]], with eigenvalues ±√4.25. The odd-parity block has eigenvalues ±0.5, so the full ground energy is −√4.25≈−2.061553. The notebook diagonalizes the full 4×4 matrix and separately searches 2,001 ansatz angles. This ansatz can represent the ground state, but the finite grid usually leaves a small positive optimization error.

This is an exact-expectation variational calculation. It has no shot noise, hardware execution or noisy optimizer. A more expressive ansatz is not automatically easier to train. Separate ansatz limitation, optimization error, finite-shot uncertainty and hardware bias when interpreting a real VQE result.

## Predict

The computed energy should lie at or above −√4.25 and within 2×10⁻⁵ for the default grid. Coarsening the grid changes optimization error without changing the Hamiltonian.
