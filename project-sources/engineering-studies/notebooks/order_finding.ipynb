# Separate order finding from factoring post-processing

Prerequisites: Modular arithmetic, greatest common divisors and even orders.

For coprime a and N, the multiplicative order is the least positive r with aʳ≡1 mod N. If r is even and a^(r/2) is neither +1 nor −1 modulo N, gcd(a^(r/2)−1,N) and gcd(a^(r/2)+1,N) can reveal nontrivial factors. The conditions are part of the argument, not implementation details to skip.

The companion finds r by classical repeated modular multiplication and performs this post-processing. For a=2,N=15 the powers are 2,4,8,1, so r=4; gcd(3,15)=3 and gcd(5,15)=5. It also handles the easy case where gcd(a,N)>1 before order finding.

Shor's quantum algorithm uses a quantum order-finding procedure and classical reconstruction. This notebook implements the arithmetic bridge, not full Shor, reversible modular exponentiation, continued-fraction recovery from sampled phases, or a cryptographically meaningful computation. A demonstration factoring 15 does not establish practical cryptographic resource requirements.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Classical precheck | gcd(a,N) | May immediately find a factor |
| Order | Smallest r with aʳ≡1 | Computed classically here |
| Post-processing | Two gcd calculations | Requires nontrivial even-order case |
