# Verify a quantum Fourier transform as a unitary

Prerequisites: Complex exponentials, orthogonality and matrix conjugate transpose.

For N=2ⁿ, the chosen forward convention is F[y,x]=exp(2πixy/N)/√N. The inverse is F†, with the opposite sign. The transform maps each basis vector to a uniform-magnitude superposition with a particular phase pattern. Measuring a transformed basis state therefore yields uniform probabilities even though its phases distinguish it from other transformed states.

The notebook constructs the full matrix for up to eight qubits and checks F†F≈I. It transforms |1⟩, then applies the inverse and verifies recovery of |1⟩. This dense classical implementation scales quadratically in N; it is a reference calculation rather than an efficient quantum circuit implementation.

Circuit diagrams can use reversed wire order and different bit significance. A successful round trip alone will not detect a consistently reversed convention, so write down the basis order and test individual phase entries as well. QFT becomes useful inside interference algorithms; measuring its amplitudes does not return a full classical Fourier spectrum in one shot.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Forward convention | exp(+2πixy/N)/√N | State the sign |
| Inverse | Conjugate transpose | Cancels the transform |
| Measurement | |amplitude|² | Does not expose all phases |

## Predict

With three qubits N=8, each probability after F|1⟩ is 1/8. Inverting must return the original state with residual below 10⁻¹².

```python
from pathlib import Path
import sys, math
base=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(base))
import quantum_models as m
```

```python
result=m.qft(3)
assert max(abs(p-.125) for p in result['probabilities'])<1e-12
assert result['unitary_residual']<1e-12
assert result['roundtrip_residual']<1e-12
print(result)
```
