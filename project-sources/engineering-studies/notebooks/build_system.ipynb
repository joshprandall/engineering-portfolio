# Integrate tools with acceptance checks

Prerequisites: The typed-agent and checkpoint/replay companions.

Building an assistant begins with a bounded task and observable acceptance criteria. This integration fixture combines a permitted tool effect, replay after an injected failure, denial without the capability, and rejection of a conflicting idempotency key. All three checks must pass before adding any new tool boundary.

The model that proposes a tool call should not own permission validation, schema validation or the effect ledger. Those belong in inspectable application code. This package deliberately uses deterministic requests so the execution guarantees can be tested without conflating model behavior with tool reliability. A future model adapter needs its own evaluation, budgets, timeout and cancellation behavior, and must not place server secrets in browser source.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Proposal | Structured tool request | Untrusted until validated |
| Execution | Schema + permission + idempotency | Application-owned boundary |
| Acceptance | Observable success and failure cases | Evidence before expansion |
