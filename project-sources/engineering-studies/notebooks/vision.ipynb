# Inspect a spatial filter and its boundary assumptions

Prerequisites: Arrays, convolution/correlation and image sampling.

A local image filter combines nearby intensities with a spatial kernel. The notebook applies a horizontal Sobel-style correlation kernel to a synthetic 7×7 vertical step edge. Valid correlation produces a 5×5 output because it uses only positions where the entire 3×3 kernel fits.

The kernel sums to zero, so adding a constant brightness offset leaves the interior response unchanged. This algebraic property does not imply robustness to contrast scaling, clipping, noise, rotation or boundary padding. The operation is explicitly correlation: the kernel is not flipped as in the mathematical definition of convolution. Deep-learning libraries often use this convention while naming the operation convolution.

An edge response is not object recognition. A trained classifier additionally needs a dataset, target labels, split policy and task-level evaluation. Preserve this simple filter as a diagnostic baseline when studying learned features.
