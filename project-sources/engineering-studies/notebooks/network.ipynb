# Separate heralding probability from conditional fidelity

Prerequisites: Bell states, mixed states and conditional probability.

A Werner link can be written ρ=w|Φ+⟩⟨Φ+|+(1−w)I/4, with Bell fidelity (1+3w)/4. Under the idealized swapping assumptions in this fixture, two independent identical links give visibility w² and conditional fidelity (1+3w²)/4. A separate simplified heralding model has success probability η²/2.

Conditional quality and probability of success answer different questions. Reporting fidelity only after successful heralding does not reveal link throughput or waiting-time effects. This fixture omits memory decay, scheduling, purification, dark counts and imperfect operations, so it is an algebraic networking exercise rather than an engineering rate forecast.

## Model structure

| Element | Model | Interpretation |
| --- | --- | --- |
| Input state | Werner visibility w | Bell fidelity |
| Swap quality | Visibility w² | Independent ideal links |
| Heralding | η²/2 | Separate conditional success model |
